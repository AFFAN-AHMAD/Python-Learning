```python

from itertools import (permutations)
def get_all_permutations(n):
    listing = [i+1 for i in range(n)]
    return permutations(listing)
```

```python
def get_the_str_list(n):
    perm_list = get_all_permutations(n)
    string_list = []
    for perm_tuple in perm_list:
        integ = ""
        for num in perm_tuple:
            integ+=str(num)
        string_list.append(integ)
    return string_list
```

```python
def get_the_code_list(n):
    return filter(lambda x: int(x) % n==0, get_the_str_list(n))
print(list(get_the_code_list(3)))
```
