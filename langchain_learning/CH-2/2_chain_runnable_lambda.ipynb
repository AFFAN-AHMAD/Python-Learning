```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm_kimi = ChatOpenAI(
    base_url='https://api.moonshot.ai/v1',
    api_key=os.environ.get('KIMI_API_KEY'),
    model='kimi-k2.6'
)
```

```python
template = ChatPromptTemplate.from_messages([
    ('system', 'You are a gen-z assistant who knows what to speak when'),
    ('user', 'Let me know the topic: {topic}')
])
```

```python
str_parser = StrOutputParser()
```

```python
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text: str) -> dict:
    return {
        'text': text
    }

runnable_dict_maker = RunnableLambda(dictionary_maker)
```

```python
prompt_post = ChatPromptTemplate.from_messages([
    ('system', 'You are a social media post generator, who writes very greate engaging post'),
    ('user', 'Create a post for the following text for Linkedin: {text}')
])
```

```python
chain = template | llm_kimi | str_parser | runnable_dict_maker | prompt_post | llm_kimi | str_parser
chain.invoke('What is the capital of France')
```
