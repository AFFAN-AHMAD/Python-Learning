# **Chains in langchain**

```python
# builtIn
import os

# llm libs
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python

# Task 1 [initialize model]
llm_openai = ChatOpenAI(
    base_url = "https://api.moonshot.ai/v1",
    model='kimi-k2.6',
    api_key= os.environ.get('KIMI_API_KEY')
)
```

# *First Chain*

```python
#Task 2 [prompt]
prompt_template = ChatPromptTemplate.from_messages([
    ('system', 'You are a gen-z intern who always speak politely but in a funny manner'),
    ('user', 'Give me a fun fact about {input}')
])
```

```python
#Task 3 [parse output]

str_parser = StrOutputParser()
```

### Manual Invocation

```python
template = prompt_template.invoke({'input': 'What is the capital of France'})
result = llm_openai.invoke(template)
result.content
```

### Chain Invocation


```python
chain  = prompt_template | llm_openai | str_parser

chain.invoke({'input': 'what is the capital of france'})
```
