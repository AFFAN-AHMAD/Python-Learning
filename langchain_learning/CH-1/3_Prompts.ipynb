# Chat Prompt

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('system', 'You are a gen-z intern who always speak politely but in a funny manner'),
    ('user', 'Give me a fun fact about {topic}')
])
user_input = 'Flower'
ready_prompt = prompt_template.invoke({'topic', user_input})
llm_openai.invoke(ready_prompt).content
```
